```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
```

```python
df_filtered.columns
```

```python
feature_columns = ['_wdire', '_dewptm', '_fog', '_hail', '_hum',
       '_pressurem', '_rain', '_snow', '_tempm', '_thunder', '_tornado',
       '_vism', '_wdird', '_wspdm', 'year', 'month', '_conds']
```

```python
df_final = df_filtered[feature_columns]
```

```python
df_final.head()
```

```python
df_final.dtypes
```

```python
df_final._wdire.value_counts()
```

```python
wdire_dummies = pd.get_dummies(df_final["_wdire"])
```

```python
df_final = pd.concat([wdire_dummies, df_final], axis=1)
```

```python
df_final.head()
```

```python
df_final.columns
```

```python
df_final.drop("_wdire", inplace=True, axis=1)
```

```python
df_final.columns
```

```python
X = df_final.iloc[:, 0:-1].values
X.shape
```

```python
y = df_final.iloc[:, -1].values
```

```python
label_encoder = LabelEncoder()
```

```python
y = label_encoder.fit_transform(y)
```

```python
y.shape
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=.3, random_state=0)
```

```python
print("Shape of X_train", X_train.shape)
print("Shape of X_test", X_test.shape)
print("Shape of y_train", y_train.shape)
print("Shape of y_test", y_test.shape)
```

```python
from sklearn.tree import DecisionTreeClassifier
clf = DecisionTreeClassifier(criterion="entropy", random_state=0)
```

```python
clf.fit(X_train, y_train)
```

```python
y_pred = clf.predict(X_test)
```

```python
y_pred
```

```python
from sklearn import metrics
import matplotlib.pyplot as plt
print("DecisionTrees's Accuracy: ", metrics.accuracy_score(y_test, y_pred))
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
regr = LinearRegression()
 
regr.fit(X_train, y_train)
print(regr.score(X_test, y_test))
```

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import train_test_split 
from sklearn.metrics import r2_score 
dtr = DecisionTreeRegressor()
dtr.fit(X_train, y_train)
pred = dtr.predict(X_test)
r2_score(y_test, pred)
```

```python
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np
```

```python
model = RandomForestRegressor(n_estimators=100, max_depth=10, random_state=42)
model.fit(X_train, y_train)

# Predict target variable using testing data
y_pred = model.predict(X_test)

# Evaluate model performance
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)
print('RMSE:', rmse)
print('R2 score:', r2)
```
